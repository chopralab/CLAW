```python

# Align column names for comparison
df_column = df["Parent_Ion"]
caitlin_column = caitlin_df["Parent Ion"]

# 1. Elements in both
common_elements = set(df_column).intersection(caitlin_column)
print("Common elements between 'Parent_Ion' and 'Parent Ion':")
print(common_elements)

# 2. Elements only in df
unique_to_df = set(df_column) - set(caitlin_column)
print("\nElements unique to 'Parent_Ion' in df:")
print(unique_to_df)

# 3. Elements only in caitlin_df
unique_to_caitlin_df = set(caitlin_column) - set(df_column)
print("\nElements unique to 'Parent Ion' in caitlin_df:")
print(unique_to_caitlin_df)

# 4. All elements combined
all_elements = set(df_column).union(caitlin_column)
print("\nAll elements from both columns:")
print(all_elements)
```

```python
# Assuming values in df and caitlin_df are numeric for comparison with a tolerance

# Function to compare with tolerance
def compare_with_tolerance(values_df, values_caitlin, tolerance=0.3):
    matched = set()
    unique_to_df = set()
    unique_to_caitlin = set(values_caitlin)

    for val in values_df:
        matched_within_tolerance = False
        for comp_val in values_caitlin:
            if abs(val - comp_val) <= tolerance:
                matched.add(comp_val)
                matched_within_tolerance = True
        if not matched_within_tolerance:
            unique_to_df.add(val)
    unique_to_caitlin -= matched
    return matched, unique_to_df, unique_to_caitlin
```

```python
matched_values, unique_to_df_values, unique_to_caitlin_values = compare_with_tolerance(
    df_column, caitlin_column, tolerance=0.5
)

```

```python
# Print results with tolerance
print("Matched values with a tolerance of ±0.3:")
print(matched_values)

print("\nValues unique to df considering tolerance:")
print(unique_to_df_values)

print("\nValues unique to caitlin_df considering tolerance:")
print(unique_to_caitlin_values)
```

```python
comparison_results = pd.DataFrame({
    "Matched Values (±0.3)": pd.Series(list(matched_values)),
    "Unique to sanjay_df (±0.3)": pd.Series(list(unique_to_df_values)),
    "Unique to caitlin_df (±0.3)": pd.Series(list(unique_to_caitlin_values))
})
```

```python
import pandas as pd
import numpy as np

def analyze_mass_tolerance(df, tolerance=0.3):
    """
    Analyze how many values fall within a specified mass tolerance range.
    
    Parameters:
    df (pandas.DataFrame): DataFrame containing mass spectrometry data
    tolerance (float): Mass tolerance range (±) in Da/mz units
    
    Returns:
    pandas.DataFrame: Summary of groups within tolerance ranges
    """
    # Sort the dataframe by Parent_Ion values
    sorted_df = df.sort_values('Parent_Ion')
    
    # Initialize lists to store group information
    groups = []
    current_group = []
    reference_mass = None
    
    # Group masses within tolerance
    for mass in sorted_df['Parent_Ion']:
        if reference_mass is None:
            reference_mass = mass
            current_group = [mass]
        elif abs(mass - reference_mass) <= tolerance:
            current_group.append(mass)
        else:
            groups.append(current_group)
            current_group = [mass]
            reference_mass = mass
    
    # Add the last group
    if current_group:
        groups.append(current_group)
    
    # Create summary DataFrame
    summary = pd.DataFrame({
        'Reference_Mass': [group[0] for group in groups],
        'Group_Size': [len(group) for group in groups],
        'Mass_Range': [f"{min(group):.2f} - {max(group):.2f}" for group in groups],
        'Mass_Spread': [max(group) - min(group) for group in groups],
        'Values': [', '.join([f"{x:.2f}" for x in group]) for group in groups]
    })
    
    return summary
```

```python
import pandas as pd
import numpy as np

def analyze_mass_tolerance(grouped_df, tolerance=0.3):
    """
    Analyze how many values fall within a specified mass tolerance range for a grouped DataFrame.
    Include Lipid and Species values for each mass group.
    
    Parameters:
    grouped_df (pandas.DataFrame): Grouped DataFrame containing mass spectrometry data
    tolerance (float): Mass tolerance range (±) in Da/mz units
    
    Returns:
    pandas.DataFrame: Summary of groups within tolerance ranges including Lipid and Species values
    """
    # Create dictionaries of mass to lipid and species mapping
    mass_lipid_dict = dict(zip(grouped_df['Parent_Ion'], grouped_df['Lipid']))
    mass_species_dict = dict(zip(grouped_df['Parent_Ion'], grouped_df['Species']))
    
    # Sort the masses
    sorted_masses = np.sort(grouped_df['Parent_Ion'].values)
    
    # Initialize lists to store group information
    groups = []
    current_group = []
    current_lipids = []
    current_species = []
    reference_mass = None
    
    # Group masses within tolerance
    for mass in sorted_masses:
        if reference_mass is None:
            reference_mass = mass
            current_group = [mass]
            current_lipids = [mass_lipid_dict[mass]]
            current_species = [mass_species_dict[mass]]
        elif abs(mass - reference_mass) <= tolerance:
            current_group.append(mass)
            current_lipids.append(mass_lipid_dict[mass])
            current_species.append(mass_species_dict[mass])
        else:
            groups.append((current_group, current_lipids, current_species))
            current_group = [mass]
            current_lipids = [mass_lipid_dict[mass]]
            current_species = [mass_species_dict[mass]]
            reference_mass = mass
    
    # Add the last group
    if current_group:
        groups.append((current_group, current_lipids, current_species))
    
    # Create summary DataFrame
    summary = pd.DataFrame({
        'Reference_Mass': [group[0][0] for group in groups],
        'Group_Size': [len(group[0]) for group in groups],
        'Mass_Range': [f"{min(group[0]):.2f} - {max(group[0]):.2f}" for group in groups],
        'Mass_Spread': [max(group[0]) - min(group[0]) for group in groups],
        'Mass_Values': [', '.join([f"{x:.2f}" for x in group[0]]) for group in groups],
        'Lipid_Values': [' | '.join(group[1]) for group in groups],
        'Species_Values': [' | '.join(group[2]) for group in groups]
    })
    
    return summary

# Example usage:
# results = analyze_mass_tolerance(grouped_df, tolerance=0.3)
```

```python
import pandas as pd
import numpy as np

def filter_species_range(results_df):
    """
    Filter out groups where all Species values are either:
    - all above 24:X OR
    - all below 12:X
    
    Parameters:
    results_df (pandas.DataFrame): DataFrame from analyze_mass_tolerance function
    
    Returns:
    pandas.DataFrame: Filtered DataFrame excluding groups where all species are outside desired range
    """
    def check_species_range(species_str):
        # Split species into individual values
        species_list = [s.strip() for s in species_str.split('|')]
        
        # Track if we find any values in our desired range
        all_high = True
        all_low = True
        
        # Parse each species value
        for species in species_list:
            try:
                # Extract the number before the colon
                num = int(species.split(':')[0])
                if num <= 24:  # Check if not too high
                    all_high = False
                if num >= 12:  # Check if not too low
                    all_low = False
                    
                # If we've found at least one value in range, we can stop checking
                if not all_high and not all_low:
                    break
                    
            except (ValueError, IndexError):
                # If we can't parse the number, assume it's in range
                all_high = False
                all_low = False
                
        # Return True to keep rows where not all species are too high or too low
        return not (all_high or all_low)

    # Apply the filter
    filtered_df = results_df[results_df['Species_Values'].apply(check_species_range)].copy()
    
    return filtered_df

# Example usage:
# results = analyze_mass_tolerance(grouped_df, tolerance=0.3)
filtered_results = filter_species_range(results)
```

```python
import pandas as pd
import numpy as np

def compare_and_filter_masses(caitlin_df, filtered_results, tolerance=0.3):
    """
    Compare Parent Ion masses between caitlin_df and filtered_results within a tolerance range
    and return both matching and non-matching results.
    
    Parameters:
    caitlin_df (pandas.DataFrame): DataFrame containing 'Parent Ion' column
    filtered_results (pandas.DataFrame): DataFrame containing 'Reference_Mass' column
    tolerance (float): Mass tolerance range (±) in Da/mz units
    
    Returns:
    tuple: (matching_df, non_matching_df) containing results that match/don't match caitlin_df values
    """
    def find_match(mass, reference_masses, tolerance):
        """Helper function to check if a mass has a match within tolerance"""
        return any(abs(mass - ref_mass) <= tolerance for ref_mass in reference_masses)
    
    # Get unique masses from both dataframes
    caitlin_masses = caitlin_df['Parent Ion'].unique()
    result_masses = filtered_results['Reference_Mass'].values
    
    # Create mask for matching masses
    mask = filtered_results['Reference_Mass'].apply(
        lambda x: find_match(x, caitlin_masses, tolerance)
    )
    
    # Split into matching and non-matching results
    matched_results = filtered_results[mask].copy()
    non_matched_results = filtered_results[~mask].copy()
    
    # Add information about matching to matched results
    matched_results['Matching_Caitlin_Masses'] = matched_results['Reference_Mass'].apply(
        lambda x: ', '.join([f"{m:.2f}" for m in caitlin_masses if abs(m - x) <= tolerance])
    )
    
    # Print summary
    print("\nMatching Summary:")
    print(f"Original number of masses in filtered_results: {len(filtered_results)}")
    print(f"Number of masses in caitlin_df: {len(caitlin_masses)}")
    print(f"Number of matching masses (within ±{tolerance} tolerance): {len(matched_results)}")
    print(f"Number of non-matching masses: {len(non_matched_results)}")
    
    # Also find caitlin masses that don't match any in filtered_results
    unmatched_caitlin_masses = [
        mass for mass in caitlin_masses 
        if not find_match(mass, filtered_results['Reference_Mass'], tolerance)
    ]
    
    print(f"\nNumber of caitlin masses without matches: {len(unmatched_caitlin_masses)}")
    if unmatched_caitlin_masses:
        print("Unmatched Caitlin masses:", ', '.join([f"{m:.2f}" for m in unmatched_caitlin_masses]))
    
    return matched_results, non_matched_results

# Example usage:
# matching_df, non_matching_df = compare_and_filter_masses(caitlin_df, filtered_results, tolerance=0.3)
```

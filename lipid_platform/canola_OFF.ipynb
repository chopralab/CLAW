```python
import os
import glob
import pandas as pd

# Define the directory that contains the CSV files
csv_dir = 'Projects/NIST/analysis/OFF/off_possible/zero/csv_dir'
# Define the output directory for blank-subtracted files
output_dir = os.path.join(csv_dir, 'blank_subtracted')
os.makedirs(output_dir, exist_ok=True)

# Get a list of all CSV files in the directory
csv_files = glob.glob(os.path.join(csv_dir, '*.csv'))

# Identify the blank file (assume the filename contains 'Blank')
blank_files = [f for f in csv_files if 'Blank' in os.path.basename(f)]
if not blank_files:
    raise ValueError("No file with 'Blank' in its name found!")
elif len(blank_files) > 1:
    print("Warning: More than one 'Blank' file found. Using the first one.")

# Read the blank file into a DataFrame
blank_file = blank_files[0]
print(f"Using blank file: {os.path.basename(blank_file)}")
blank_df = pd.read_csv(blank_file)

# Create a lookup for blank intensities by Lipid.
# Assumes one row per unique Lipid.
blank_lookup = blank_df.set_index('Lipid')['OzESI_Intensity']

# Debug: show the first few items in the blank lookup
print("First few entries in the blank lookup:")
print(blank_lookup.head())

# Now iterate through all CSV files (skipping the blank file)
for file in csv_files:
    if file == blank_file:
        continue  # Skip the blank file

    print(f"\nProcessing file: {os.path.basename(file)}")
    df = pd.read_csv(file)

    # Optionally, if you want to filter to a specific lipid (e.g. containing '18:1'):
    # df = df[df['Lipid'].str.contains('18:1')]

    # Map the blank intensity to the current file's lipids.
    df['Blank_Intensity'] = df['Lipid'].map(blank_lookup)

    # Debug: count how many lipids got a matching blank intensity
    num_matched = df['Blank_Intensity'].notna().sum()
    total_lipids = len(df)
    print(f"Matched blank intensity for {num_matched} out of {total_lipids} lipids.")

    # Optionally, show some of the mapping details for the first few lipids
    print("Sample mapping (first 5 rows):")
    print(df[['Lipid', 'OzESI_Intensity', 'Blank_Intensity']].head())

    # Replace missing blank values with 0 if that's appropriate
    df['Blank_Intensity'] = df['Blank_Intensity'].fillna(0)

    # Subtract the blank intensity from the original intensity
    df['OzESI_Intensity_BlankSubtracted'] = df['OzESI_Intensity'] - df['Blank_Intensity']

    # Debug: display the result of subtraction for the first few rows
    print("After subtraction (first 5 rows):")
    print(df[['Lipid', 'OzESI_Intensity', 'Blank_Intensity', 'OzESI_Intensity_BlankSubtracted']].head())

    # Reorder columns so that OzESI_Intensity_BlankSubtracted is right after OzESI_Intensity
    cols = list(df.columns)
    # Remove the new column from the list (if it exists) so we can reinsert it
    if 'OzESI_Intensity_BlankSubtracted' in cols:
        cols.remove('OzESI_Intensity_BlankSubtracted')
    # Find the index of OzESI_Intensity and insert the new column right after it
    try:
        idx = cols.index('OzESI_Intensity')
    except ValueError:
        raise ValueError("Column 'OzESI_Intensity' not found in the DataFrame!")
    cols.insert(idx + 1, 'OzESI_Intensity_BlankSubtracted')
    # Reorder the dataframe columns
    df = df[cols]

    # Optionally drop the extra Blank_Intensity column if you don't need it
    # df.drop(columns=['Blank_Intensity'], inplace=True)

    # Save the new dataframe to the output directory.
    output_file = os.path.join(output_dir, os.path.basename(file))
    df.to_csv(output_file, index=False)
    print(f"Processed and saved: {output_file}")
```

```python
import matplotlib.pyplot as plt

# Function to plot the chromatogram for a specific group_by_lipid value
def plot_chromatogram(df, group_value):
    group_data = df[df['group_by_lipid'] == group_value]
    
    # Ensure the Retention_Time is sorted
    group_data = group_data.sort_values(by='Retention_Time')
    
    # Extract Lipid, Parent Ion, and Isomer values from the first row of the group
    lipid = group_data['Lipid'].iloc[0] if not group_data.empty else "Unknown Lipid"
    parent_ion = group_data['Parent_Ion'].iloc[0] if not group_data.empty else "Unknown Parent Ion"
    # isomer = group_data['Isomer'].iloc[0] if not group_data.empty else "Unknown Isomer"
    
    plt.figure(figsize=(10, 6))
    plt.plot(group_data['Retention_Time'], group_data['OzESI_Intensity'], label=f'group_by_lipid: {group_value}')
    plt.xlabel('Retention Time')
    plt.ylabel('OzESI Intensity')
    
    # Add Lipid, Parent Ion, and Isomer to the title
    plt.title(f'{group_value} | Lipid: {lipid} | Parent Ion: {parent_ion} ')
    
    plt.legend()
    plt.grid(True)
    plt.show()

# Define the specific group_by_lipid value you are interested in
specific_group_by_lipid_value = 26 # Replace with the actual value you are interested in

# Call the function to plot the chromatogram
plot_chromatogram(df_group, specific_group_by_lipid_value)
```

```python
import pymzml
import pandas as pd

# Path to the uploaded mzML file
file_path = 'SplashMRM_22Oct-TuneSampleID_SRM.mzML'

# Load the mzML file using pymzml
run = pymzml.run.Reader(file_path)

# Initialize lists to store peak information
peaks_data = []

# Debug counters
total_chromatograms = 0
peaks_found = 0

# Iterate through the chromatograms and perform peak detection
for chromatogram in run:
    total_chromatograms += 1
    if 'SRM' in chromatogram.ID:  # Checking if the chromatogram is SRM
        # Extracting the precursor m/z and product m/z from the chromatogram metadata
        precursor_mz = chromatogram.get('precursor').get('isolationWindow').get('target m/z')
        product_mz = chromatogram.ID.split('Q3=')[1].split()[0]  # Extracting product m/z from ID
        
        # Extracting time and intensity pairs from the chromatogram
        for time, intensity in chromatogram.peaks('centroided'):
            if intensity > 0:  # Only considering positive intensities as peaks
                peaks_data.append({
                    'Time': time,
                    'Intensity': intensity,
                    'Precursor m/z': precursor_mz,
                    'Product m/z': float(product_mz)
                })
                peaks_found += 1

# Display the results
peaks_df = pd.DataFrame(peaks_data)

# Print debugging information
print(f"mzML file loaded successfully.")
print(f"Total chromatograms processed: {total_chromatograms}")
print(f"Total peaks found: {peaks_found}")

if peaks_df.empty:
    print("No peaks detected across all chromatograms.")
else:
    print(peaks_df)
```

```python
import pymzml
import pandas as pd

# Path to the uploaded mzML file
file_path = 'SplashMRM_22Oct-TuneSampleID_SRM.mzML'
#file_path = '20241023_PlasmaExtract4.mzML'

# Load the mzML file using pymzml
run = pymzml.run.Reader(file_path)

# Initialize lists to store peak information
peaks_data = []

# Debug counters
total_chromatograms = 0
peaks_found = 0

# Iterate through the chromatograms and perform peak detection
for chromatogram in run:
    total_chromatograms += 1
    chromatogram_id = str(chromatogram.ID)  # Ensure chromatogram_id is a string

    # Attempt to extract Q1 and Q3 from the chromatogram's metadata (e.g., its ID)
    q1_value = None
    q3_value = None

    if 'Q1=' in chromatogram_id and 'Q3=' in chromatogram_id:
        try:
            # Extract Q1 and Q3 from the chromatogram ID string
            q1_value = float(chromatogram_id.split('Q1=')[1].split()[0])
            q3_value = float(chromatogram_id.split('Q3=')[1].split()[0])
        except (IndexError, ValueError) as e:
            print(f"Error parsing Q1 and Q3 values for chromatogram {total_chromatograms}: {e}")
            continue

    # If Q1 and Q3 were successfully extracted, process the chromatogram peaks
    if q1_value is not None and q3_value is not None:
        for time, intensity in chromatogram.peaks('raw'):  # Use 'raw' data for MRM
            if intensity > 0:  # Consider only positive intensities as peaks
                peaks_data.append({
                    'Time': time,
                    'Intensity': intensity,
                    'Q1 (Precursor m/z)': q1_value,
                    'Q3 (Product m/z)': q3_value
                })
                peaks_found += 1
    else:
        print(f"Chromatogram {total_chromatograms} does not have Q1 and Q3 information in the expected format.")

# Convert the peaks data to a DataFrame
peaks_df = pd.DataFrame(peaks_data)

# Print debugging information
print(f"mzML file loaded successfully.")
print(f"Total chromatograms processed: {total_chromatograms}")
print(f"Total peaks found: {peaks_found}")

# Display the DataFrame or indicate if no peaks were found
if peaks_df.empty:
    print("No peaks detected across all chromatograms.")
else:
    print(peaks_df)
```

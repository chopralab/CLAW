```python
import os
from pathlib import Path
import pandas as pd

def process_parquet_files(input_dir):
    """
    Process all parquet files in the input_dir:
      - For each file, keep only the top 2 OzESI_Intensity rows per (Lipid, Sample) combo.
      - Write one CSV containing all filtered rows.
      - Write separate CSVs for each unique Species value.
      - Each file gets its own subdirectory (named after the parquet filename without extension).
    
    Parameters:
        input_dir (str or Path): Path to the directory containing the parquet files.
    """
    input_dir = Path(input_dir)
    
    # Loop over all parquet files in the directory.
    for parquet_file in input_dir.glob("*.parquet"):
        print(f"Processing file: {parquet_file.name}")
        # Read the parquet file into a DataFrame.
        df = pd.read_parquet(parquet_file)
        
        # Ensure the required columns exist.
        required_columns = {'Lipid', 'Sample', 'OzESI_Intensity'}
        if not required_columns.issubset(df.columns):
            print(f"Skipping {parquet_file.name} because it lacks one or more required columns: {required_columns}")
            continue
        
        # Sort the DataFrame so that the highest OzESI_Intensity is first.
        df_sorted = df.sort_values('OzESI_Intensity', ascending=False)
        
        # Group by the combination of Lipid and Sample and take the top 2 rows from each group.
        df_top2 = df_sorted.groupby(['Lipid', 'Sample'], as_index=False).head(2)
        
        # Create an output subdirectory with the same base name as the parquet file.
        out_dir = input_dir / parquet_file.stem
        out_dir.mkdir(exist_ok=True)
        
        # Write the entire filtered DataFrame to a CSV file.
        all_csv_path = out_dir / f"{parquet_file.stem}_ALL.csv"
        df_top2.to_csv(all_csv_path, index=False)
        print(f"Written all data to {all_csv_path}")
        
        # Check that a 'Species' column exists.
        if 'Species' not in df_top2.columns:
            print(f"File {parquet_file.name} does not contain a 'Species' column. Skipping species-based CSV generation.")
            continue
        
        # For each unique species, write a separate CSV file.
        for species, df_species in df_top2.groupby('Species'):
            # Sanitize species string for use in a filename (e.g., replace ':' with '-' if needed).
            species_safe = species.replace(":", "-")
            species_csv_path = out_dir / f"{parquet_file.stem}_{species_safe}.csv"
            df_species.to_csv(species_csv_path, index=False)
            print(f"Written species data ({species}) to {species_csv_path}")

# Example usage:
if __name__ == "__main__":
    # Replace with your actual directory path.
    isomer_filter_dir = "Projects/NIST/isomer_filter"
    process_parquet_files(isomer_filter_dir)
```

```python
import matplotlib.pyplot as plt

# Function to plot the chromatogram for a specific group_by_lipid value
def plot_chromatogram(df, group_value):
    group_data = df[df['group_by_lipid'] == group_value]
    
    # Ensure the Retention_Time is sorted
    group_data = group_data.sort_values(by='Retention_Time')
    
    # Extract Lipid, Parent Ion, and Isomer values from the first row of the group
    lipid = group_data['Lipid'].iloc[0] if not group_data.empty else "Unknown Lipid"
    parent_ion = group_data['Parent_Ion'].iloc[0] if not group_data.empty else "Unknown Parent Ion"
    # isomer = group_data['Isomer'].iloc[0] if not group_data.empty else "Unknown Isomer"
    
    plt.figure(figsize=(10, 6))
    plt.plot(group_data['Retention_Time'], group_data['OzESI_Intensity'], label=f'group_by_lipid: {group_value}')
    plt.xlabel('Retention Time')
    plt.ylabel('OzESI Intensity')
    
    # Add Lipid, Parent Ion, and Isomer to the title
    plt.title(f'{group_value} | Lipid: {lipid} | Parent Ion: {parent_ion} ')
    
    plt.legend()
    plt.grid(True)
    plt.show()

# Define the specific group_by_lipid value you are interested in
specific_group_by_lipid_value = 324  # Replace with the actual value you are interested in

# Call the function to plot the chromatogram
plot_chromatogram(df_group, specific_group_by_lipid_value)
```

```python
import os
from pathlib import Path

def create_project_structure(project_name, base_path='Projects'):
    """
    Creates a standardized directory structure for lipid analysis projects.
    
    Parameters:
    -----------
    project_name : str
        Name of the project directory (e.g., 'FAME', 'CT')
    base_path : str
        Path where the project directory should be created (default: 'Projects')
        
    Returns:
    --------
    project_path : Path
        Path object pointing to the created project directory
    """
    # Define the base structure
    structure = {
        'analysis': ['NP', 'OFF', 'ON'],
        'group': ['NP', 'OFF', 'ON'],
        'isomer_filter': [],  # Empty list means no subdirectories
        'match': ['NP', 'OFF', 'ON'],
        'mzml': ['OFF', 'ON'],
        'mzml_parsed': ['NP', 'OFF', 'ON'],
        'not_possible': [],
        'results': [],
        'samples': ['NP', 'OFF', 'ON']
    }
    
    # Create the project directory
    project_path = Path(base_path) / project_name
    project_path.mkdir(parents=True, exist_ok=True)
    
    # Create the directory structure
    for dir_name, subdirs in structure.items():
        # Create main directory
        main_dir = project_path / dir_name
        main_dir.mkdir(exist_ok=True)
        
        # Create subdirectories if any
        for subdir in subdirs:
            sub_path = main_dir / subdir
            sub_path.mkdir(exist_ok=True)
    
    print(f"Created project structure at: {project_path}")
    return project_path

def print_directory_tree(path, level=0, max_level=2):
    """
    Prints a directory tree structure up to specified level.
    
    Parameters:
    -----------
    path : Path or str
        Starting path to print from
    level : int
        Current level in the tree (used internally)
    max_level : int
        Maximum depth to print
    """
    path = Path(path)
    if level > max_level:
        return
        
    prefix = '│   ' * (level - 1) + '├── ' if level > 0 else ''
    print(f"{prefix}{path.name}")
    
    try:
        contents = sorted(list(path.iterdir()))
        for i, item in enumerate(contents):
            if item.is_dir():
                print_directory_tree(item, level + 1, max_level)
    except PermissionError:
        print(f"Permission denied for {path}")

# Example usage in Jupyter notebook:
```

```python

# Create a new project structure
new_project = create_project_structure('AMP_V2')

# Print the directory tree
print_directory_tree(new_project)
```
